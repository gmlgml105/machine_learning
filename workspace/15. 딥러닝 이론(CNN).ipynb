# CNN(Convolutional Neural Network)
-  CNN(합성곱 신경망)은 이미지의 위치 정보를 유지하면서 시각적 특징을 효과적으로 추출하기 위해 개발된 딥러닝 아키텍처입니다.
- 입력 데이터를 평탄화하여 1차원으로 펼치지 않고, 2차원/3차원 공간 구조를 있는 그대로 유지하면서 합성곱(Convolution) 연산을 수행합니다.
- 필터(Filter)를 활용해 이미지의 거칠고 미세한 선, 테두리, 질감 등 주요 특징 맵(Feature Map)을 자동으로 뽑아내며 입력 변화에 유연하게 대응합니다.
- 추출된 공간 특징들은 풀링(Pooling)층을 거치며 압축된 후, 마지막 완전연결층(Fully Connected Layer, 우리가 알고 있는 딥러닝)으로 전달되어 개, 고양이 등 최종 클래스를 분류하게 됩니다.
- 즉 CNN은 사람의 눈 처럼 이미지 분류, 객체 탐지, 영상 처리 등 다양한 분야에서 높은 성능을 발휘하며, 이미지의 공간적 관계를 보존하면서 학습할 수 있다는 점에서 강점을 가진다.
- 멀티모달의 시각적인 부분을 담당한다.

## 합성곱 계층 (Convolutional Layer)
- CNN의 핵심 구성 요소로, 이미지의 공간적 구조를 유지하면서 중요한 특징을 추출하는 역할을 합니다. 작은 크기의 필터(커널)가 이미지를 따라 이동(스캔)하며 일부분씩 합성곱 연산을 수행하며, 이 과정에서 필터는 가장자리, 텍스처 등 특정 패턴을 감지하도록 학습됩니다. 스캔 결과로 생성되는 특징 맵(Feature Map)은 이미지의 핵심 정보가 강조된 새로운 데이터 표현이며, 이를 다음 계층으로 전달하여 레이어가 깊어질수록 더욱 복잡하고 High-level인 패턴을 학습할 수 있게 합니다.

## 합성곱 연산 (Convolution Operation)
- 합성곱 계층 내부에서 실제로 일어나는 수학적 계산 과정입니다. 입력 이미지의 일정 영역과 필터(커널)의 대응되는 격자값을 1:1로 곱한 뒤, 그 결과를 모두 더하는 '원소별 곱의 합(Element-wise Multiplication & Sum)' 방식으로 이루어집니다. 필터가 이미지 전체를 지정된 간격(Stride)만큼 슬라이딩하면서 이 연산을 반복하여 단일 수치로 이루어진 특징 맵의 한 칸 한 칸을 채워나갑니다. 이 연산을 통해 이미지 내 특정 패턴이 존재하는 위치에서 높은 반응값(큰 숫자)을 출력하게 됩니다.
- 커널의 크기는 3x3, 5x5, 7x7, 9x9 등 정방향 크기를 사용하나 3x3을 가장 많이 사용한다.
  
- ![](https://velog.velcdn.com/images%2Fgaram%2Fpost%2F69584689-6b3f-4610-9963-5db0570300e3%2Fconv15.png)

### 연산식
- 같은 자리에 있는 포지션 끼리의 값을 곱하고 모두 더한다.
- (1x1) + (2x0) + (3x1) + (2x1) + (1x0) + (0x1) + (3x0) + (0x1) + (1x0) = 6

### 연산식을 뽑으면 특징을 뽑아내며 출력크기가 작아진다.
- 커널은 가중치의 역할을 수행한다. 

## 스트라이드(Stride)
- 합성곱 연산에서 필터(커널)가 입력 데이터(이미지) 위에 이동하는 간격을 의미합니다. 기본적으로 스트라이드 값이 1이면 필터가 한 칸씩 움직이며,  
  모든 위치에서 연산을 수행합니다. 스트라이드 값이 2이상이면 필터가 더 큰 간격으로 이동하므로, 생성되는 출력 크기가 작아지고 연산량도 줄어듭니다.  
  스트라이드를 조절하면 특징 맵의 크기를 조정할 수 있어, 모델의 계산 효율성을 높이거나 더 큰 영역의 정보를 한 번에 처리할 수 있습니다.  
  다만 스트라이드가 너무 크면 중요한 세부 정보가 손실될 수 있으므로 적절한 값을 선택하는 것이 중요합니다.

## 패딩(padding)
- 합성곱 연산에서 입력 데이터(이미지)의 가장자리에 값을 추가하여 출력 크기를 조정하거나 경계 부분의 정보 손실을 방지하는 기법입니다.  
  일반적으로 가장 자리에 0을 추가하는 제로 패딩(Zero Padding)이 많이 사용됩니다. 패딩은 두 가지 주요 목적이 있습니다.  
  첫째 입력 데이터 크기를 유지하여 출력 크기를 줄이지 안혹 처리할 수 있게 하며(예: "same" 패딩), 경계 부분의 정보를 더 많이 학습할 수 있게 합니다.  
  패딩을 사용하지 않으면("valid" 패딩) 합성곱 연산이 진행될수록 데이터 크기가 줄어드는 문제가 발생할 수 있습니다.  
  이를 통해 모델이 공간 정보를 더 잘 학습할 수 있도록 돕습니다.
- 합성곱은 Convolution이후에 출력값에 적용된다.

## 풀링(Pooing)
- CNN에서 특징 맵의 크기를 줄이고, 중요한 정보를 요약해서 계산 효율성을 높이는 과정입니다.
  주로 사용되는 방법은 최대 풀링(Max Pooling)과 평균 풀링(Average Pooling)으로 최대 풀링은 작은 영역에서 가장 큰 값을 선택해 주요 특징을 강조하고,
  평균 풀링은 영역의 평균 값을 계산하여 전체적인 정보를 요약합니다. 풀링은 데이터 크기를 줄이면서도 중요한 특징을 유지하고 모델이 공간적 위치 변화에 덜 민감해지도록 만들어
  일반화 성능을 높입니다. 이를 통해 연산량을 줄이고 과적합(overfitting)을 방지하는 데 도움을 줍니다.
- max pooing stride가 기본 값 2이기 때문에 이미지가 확 줄어든다.

## DNN과 CNN(딥러닝과 CNN의 차이)
### DNN의 가중치와 편향
- DNN은 완전 연결 계층(fully connected layer)으로 구성되어 있어, 각 입력 노드가 모든 출력 노드와 연결됩니다. 이 때문에 입력 노드의 수와 출력 노드의 수가 비례하여 매우 많은 가중치(weight)가 필요합니다. 각 연결마다 고유한 가중치를 학습하므로, 같은 값이라도 위치가 다르면 별도의 가중치를 사용합니다. 모든 노드에 대해 별도로 편향(bias)이 존재하며, 가중치와 함께 학습됩니다.

### CNN의 가중치와 편향
- CNN은 합성곱 계층은 작은 크기의 필터(커널)을 사용하여 입력 데이터의 특정 영역만 처리합니다. 필터는 여러 위치에서 동일하게 적용되며, 이를 가중치 공유(weight sharing)이라고 합니다. 즉 필터에 사용되는 가중치는 입력 데이터의 모든 영역에서 반복적으로 사용되므로, 학습해야 할 가중치 수가 크게 줄어듭니다. 각 필터마다 하나의 편향(bias)이 있으며, 필터가 어디에 적요오디든 동일하게 사용됩니다. 합성곱 신경망에도 편항(bias)을 추가할 수 있습니다. 만약, 편향을 사용한다면 커널을 적용한 뒤에 더해집니다. 편향은 하나의 값만 존재하며, 커널이 적용된 결과의 모든 원소에 더해집니다.
- 즉 필터(커널)이 가중치이므로, 3x3 만큼의 커널만 가지고 움직이면서 학습하기 때문에 딥러닝에 비해 학습량이 작다.

## 다수의 채널 합성곱 연산
- 다수의 채널을 가진 경우(예: 컬러 이미지는 3개의 채널, RGB로 구성된다)  
  합성곱 연산은 각 채널별로 개별적으로 진행된 후 합산되어 최종 결과를 도출한다.  
  필터의 각 채널과 입력 이미지의 대응 채널에 대해 합성곱 연산을 수행합니다. 예를 들어, 필터의 R채널 부분과 이미지의 R채널 부분에서 합성곱 연산을 수행합니다.  
  동일하게 G채널과 B채널에서도 각각 수행합니다. 나온 합성곱의 결과를 모두 더해서 최종 값 하나를 만듭니다.  
  필터는 설정된 스트라이드에 따라 이미지 위를 이동하며 위 과정을 반복합니다.  
  최종적으로 각 위치에서 계산된 결과들이 모여 하나의 출력 채널(feature map)을 만듭니다.  

## **CNN을 구성하는 레이어**
- CNN은 이미지를 단계적으로 처리하여 결과를 예측합니다.

### 1. 입력 레이어(Input Layer)
- 이미지를 숫자 행렬(32 x 32 x 3(rgb))로 네트워크에 전달

### 2. 합성곱 레이어(Convolution Layer) 
- 작은 필터(예: 3x3)을 사용해 중요한 특징을 찾아낸다.
- 필터를 여러 개 사용해 여러 가지 특징을 찾아냅니다.

### 3. 활성화 함수(ReLU) 사용
- 음수 값을 0으로 바꿔 비선형을 추가하고, 모델이 복잡한 패턴을 학습할 수 있도록 돕습니다

### 4. 풀링 레이어(Pooling Layer)
- 이미지 크기를 줄이고, 중요한 정보만 요약하여 일반화 성능을 끌어올린다.
- 예: 2x2 풀링은 4개의 값 중 가장 큰 값을 선택(max Pooling)

### 5. 완전 연결 계층(Fully Connected Layer == Dense Layer)
- 이미지의 모든 특징을 하나의 벡터로 평평하게 펼쳐 최종 클래스를 예측합니다
- 예: 10개의 클래스(고양이, 개)등에 대한 확률을 게산합니다.

### 6. 출력 레이어(Output Layer)
- 소프트맥스(Softmax) 같은 함수를 사용해 각 클래스에 대한 확률을 출력합니다.

```python
inputs = torch.Tensor(1, 1, 28, 28)
print(inputs.shape)
```

## 1. Convolution Layer

```python
# in_channels: 1 -> 이미지 1개
# out_channels: 32 -> 특성 32개로 추출(마음대로)
# kernel_size: 3 -> 3x3(default)
# stride: 1 -> 1칸 슬라이딩(default)
# padding: 테두리 또는 모서리 특징 추출시 데이터의 손실을 방지하기 위해서 사용하는 값 -> zero-padding(default)
nn.Conv2d?
```

```python
conv1 = nn.Sequential(
    nn.Conv2d(in_channels=1, out_channels=32, kernel_size=3, padding="same"),
    nn.ReLU() # 활성화 함수 (선형 -> 비선형)
)

out = conv1(inputs)
print(out.size())
```

```python
pooling1 = nn.MaxPool2d(kernel_size=2) # 14 x 14 (커널의 개수에 따라서 줄어든다!)
out = pooling1(out)

print(out.size())
```

```python
conv2 = nn.Sequential(
    nn.Conv2d(in_channels=32, out_channels=64, kernel_size=3,padding="same"),
    nn.ReLU()
)

out = conv2(out)
print(out.size())
```

```python
pooling2 = nn.MaxPool2d(kernel_size=2) # 7 x 7 (커널의 개수에 따라서 줄어든다!)
out = pooling2(out)

print(out.size())
```
